```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data load karein (apna sahi file path yahan dalein)
df = pd.read_csv('/content/hotel_booking.csv')
```

```python
#Dataset shape
df.shape
```

```python
#Dataset Info
df.info()
```

```python
#Summary Statistics
df.describe(include='all')
```

**Step 2: Column-by-Column Analysis, Cleaning, and Visualizations**

```python
#1. Target Variable: is_canceled
print(df['is_canceled'].value_counts(dropna=False))

plt.figure(figsize=(6, 4))
sns.countplot(x='is_canceled', data=df, palette='Set2')
plt.title('Booking Cancellation Distribution (0 = No, 1 = Yes)')
plt.show()
```

```python
# 2. Hotel Type: hotel
print(df['hotel'].value_counts(dropna=False))
plt.figure(figsize=(6, 4))
sns.countplot(x='hotel', data=df, palette='Pastel1')
plt.title('Hotel Type Distribution')
plt.show()
```

```python
# 3. Lead Time: lead_time
# Negative values clear karein agar hain to
df = df[df['lead_time'] >= 0]

plt.figure(figsize=(10, 4))
sns.histplot(df['lead_time'], bins=50, kde=True)
plt.title('Lead Time Distribution')
plt.show()

plt.figure(figsize=(10, 2))
sns.boxplot(x=df['lead_time'])
plt.title('Lead Time Outliers')
plt.show()
```

```python
#4. Arrival Date Columns: arrival_date_year, arrival_date_month, arrival_date_week_number, arrival_date_day_of_month
print(df[['arrival_date_year', 'arrival_date_month']].value_counts())

# Month standard order ke liye plot
plt.figure(figsize=(12, 4))
sns.countplot(x='arrival_date_month', data=df, order=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'])
plt.xticks(rotation=45)
plt.title('Arrivals by Month')
plt.show()
```

```python
#5. Stay Duration: stays_in_weekend_nights aur stays_in_week_nights
df['total_stay'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

# Invalid entries check karein (stay hi nahi kiya)
invalid_stays = df[df['total_stay'] == 0]
print(f"Bookings with 0 nights stay: {len(invalid_stays)}")

# Filter out 0 stay if necessary
df = df[df['total_stay'] > 0]

plt.figure(figsize=(10, 4))
sns.countplot(x='total_stay', data=df[df['total_stay'] < 15]) # Looking at stays less than 15 days for clarity
plt.title('Total Stay Duration (Nights)')
plt.show()
```

```python
# 6. Guests Demographics: adults, children, babies
# Impute missing values
df['children'] = df['children'].fillna(0).astype(int)
df['babies'] = df['babies'].fillna(0).astype(int)
df['adults'] = df['adults'].astype(int)

# 0 guests wali rows remove karein
zero_guests = df[(df['adults'] == 0) & (df['children'] == 0) & (df['babies'] == 0)]
print(f"Rows with zero guests: {len(zero_guests)}")
df = df.drop(zero_guests.index)

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.countplot(x='adults', data=df, ax=axes[0], palette='Blues')
sns.countplot(x='children', data=df, ax=axes[1], palette='Oranges')
sns.countplot(x='babies', data=df, ax=axes[2], palette='Greens')
axes[0].set_title('Adults Count')
axes[1].set_title('Children Count')
axes[2].set_title('Babies Count')
plt.show()
```

```python
#Meal Type: meal
print(df['meal'].value_counts())
df['meal'] = df['meal'].replace('Undefined', 'SC')

sns.countplot(x='meal', data=df)
plt.title('Meal Type Chosen')
plt.show()
```

```python
# 8. Country Origin: country
df['country'] = df['country'].fillna('Unknown')

top_10_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(10, 4))
sns.barplot(x=top_10_countries.index, y=top_10_countries.values, palette='viridis')
plt.title('Top 10 Guest Countries')
plt.show()
```

```python
# 9. Market Segment & Distribution Channel: market_segment, distribution_channel
print(df['market_segment'].value_counts())
print(df['distribution_channel'].value_counts())

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
sns.countplot(x='market_segment', data=df, ax=axes[0])
axes[0].tick_params(axis='x', rotation=45)
axes[0].set_title('Market Segment')

sns.countplot(x='distribution_channel', data=df, ax=axes[1])
axes[1].tick_params(axis='x', rotation=45)
axes[1].set_title('Distribution Channel')
plt.show()
```

```python
# 10. History Variables: is_repeated_guest, previous_cancellations, previous_bookings_not_canceled
print(df['is_repeated_guest'].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(x='is_repeated_guest', data=df)
plt.title('Repeated Guest (0 = No, 1 = Yes)')
plt.show()
```

```python
# 11. Rooms Room Types: reserved_room_type, assigned_room_type
df['room_mismatch'] = df['reserved_room_type'] != df['assigned_room_type']
print(df['room_mismatch'].value_counts())

sns.countplot(x='room_mismatch', data=df, hue='is_canceled')
plt.title('Room Mismatch vs Cancellation')
plt.show()
```

```python
# 12. Booking Details: booking_changes, deposit_type, agent, company, days_in_waiting_list, customer_type
# Handling missing values for high-null columns
df['agent'] = df['agent'].fillna(0).astype(int)
df['company'] = df['company'].fillna(0).astype(int)

# Deposit Type Analysis
print(df['deposit_type'].value_counts())

sns.countplot(x='deposit_type', data=df, hue='is_canceled')
plt.title('Deposit Type vs Cancellation')
plt.show()
```

```python
#13. Financials: adr (Average Daily Rate)
# Negative ADR clean karein
df = df[df['adr'] >= 0]

# Check for extreme outliers
plt.figure(figsize=(10, 2))
sns.boxplot(x=df['adr'])
plt.title('ADR Outliers Identification')
plt.show()

# Removing the extreme outlier if adr > 5000 (usually a data entry error)
df = df[df['adr'] < 5000]
```

```python
#14. Special Requests: total_of_special_requests aur required_car_parking_spaces
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(x='total_of_special_requests', data=df, ax=axes[0], palette='magma')
axes[0].set_title('Special Requests Count')

sns.countplot(x='required_car_parking_spaces', data=df, ax=axes[1], palette='viridis')
axes[1].set_title('Required Car Parking Spaces')
plt.show()
```

```python
#Step 3: Correlation Matrix (Final Overview)
plt.figure(figsize=(14, 10))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Hotel Booking Features')
plt.show()
```

```python
print("Final Missing Values Check:")
print(df.isnull().sum().sum())

# Save cleaned dataset
df.to_csv('hotel_bookings_cleaned.csv', index=False)
print("Cleaned dataset successfully saved!")
```
