```python
# Let's save a couple of plots representing the key insights
# 1. Distribution of Numerical Features
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(df['ApplicantIncome'], kde=True, ax=axes[0,0], color='blue').set_title('Applicant Income Distribution')
sns.histplot(df['CoapplicantIncome'], kde=True, ax=axes[0,1], color='orange').set_title('Coapplicant Income Distribution')
sns.histplot(df['LoanAmount'].dropna(), kde=True, ax=axes[1,0], color='green').set_title('Loan Amount Distribution')
sns.boxplot(x='Loan_Status', y='ApplicantIncome', data=df, ax=axes[1,1]).set_title('ApplicantIncome vs Loan Status')
plt.tight_layout()
fig.savefig('numerical_distributions.png')
plt.close()
```

```python
# Missing values handling scheme
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
df['Dependents'] = df['Dependents'].fillna(df['Dependents'].mode()[0])
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History'].mode()[0])

df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].median())
```

# 1. **Dataset Overview**
Total Rows (Records): 1,614

Total Columns: 13

Duplicate Rows: No exact duplicate rows were found, but 76 Loan_IDs are repeated with different structural details (these duplicate entries can be cleaned later if required).

2. Column-by-Column EDA (Detailed Analysis)
Categorical Columns Analysis:
Loan_ID: This is the unique ID for each applicant. It contains 76 repeated keys where different profiles exist.

Gender: This column contains 980 Male and 621 Female records, with 13 missing values.

Married: 925 applicants are married, while 686 are unmarried. There are 3 missing values.

Dependents: Indicates the number of dependents. The distribution is: 0 (619), 1 (341), 2 (336), and 3+ (303) records. There are 15 missing values.

Education: 952 applicants are Graduates and 662 are Not Graduates. There are no missing values.

Self_Employed: 999 applicants are job holders, while 583 run their own business (self-employed). There are 32 missing values.

Credit_History: This is the most critical feature for loan approval. 994 applicants have a good credit history (1.0), while 570 have a poor credit history (0.0). There are 50 missing values.

Property_Area: Semiurban (545), Urban (549), and Rural (520) areas are almost equally distributed. There are no missing values.

Loan_Status (Target Variable): 915 loans were approved (Y) and 699 were rejected (N).

Numerical Columns Analysis:
ApplicantIncome: The minimum income is 150 and the maximum is 81,000. The median income is 7,045, which indicates that the data is right-skewed and contains some high-earning outliers.

CoapplicantIncome: The minimum is 0 and the maximum is 41,667. In many cases, the coapplicant income is 0 (Median: 2,941.5).

LoanAmount: The minimum is 9 and the maximum is 700 (in thousands). There are 22 missing values.

Loan_Amount_Term: Most of the loans are for 360 months (30 years). There are 14 missing values.
