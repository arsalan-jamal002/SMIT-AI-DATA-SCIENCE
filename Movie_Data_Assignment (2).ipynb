### **Question 21**
(Visualization) IMDB_Rating ka distribution dekhne ke liye ek neat Histogram plot karein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df['IMDB_Rating'] = pd.to_numeric(df['IMDB_Rating'], errors='coerce')
rating_data = df['IMDB_Rating'].dropna()

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

sns.histplot(rating_data, bins=15, kde=True, color="royalblue", edgecolor="black", alpha=0.7)
plt.title("Distribution of IMDB Ratings", fontsize=16, fontweight='bold', pad=15)
plt.xlabel("IMDB Rating", fontsize=12, labelpad=10)
plt.ylabel("Number of Movies (Count)", fontsize=12, labelpad=10)
plt.tight_layout()
plt.show()
```

### **Question 22**
(Visualization) Top 10 Directors aur unki total movies ka ek attractive Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

top_directors = df['Director'].dropna().value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    x=top_directors.values, 
    y=top_directors.index, 
    palette="viridis",      # Attractive color gradient (dark to light)
    edgecolor="black",
    hue=top_directors.index, # To avoid future warnings
    legend=False
)
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=8, fontsize=11, fontweight='bold')

plt.title("Top 10 Directors by Total Number of Movies", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Number of Movies", fontsize=12, labelpad=10)
plt.ylabel("Director Name", fontsize=12, labelpad=10)

plt.xlim(0, max(top_directors.values) + 2)
plt.tight_layout()
plt.show()
```

### **Question 23**
(Visualization) IMDB_Rating aur Gross revenue ke darmiyan ek Scatter Plot banayein taake dono ka relation dikhe.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

if 'Gross' in df.columns:
    df['Gross'] = df['Gross'].astype(str).str.replace('$', '').str.replace(',', '')
    df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')

df['IMDB_Rating'] = pd.to_numeric(df['IMDB_Rating'], errors='coerce')

plot_data = df[['IMDB_Rating', 'Gross']].dropna()

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

scatter = sns.scatterplot(
    x='IMDB_Rating', 
    y='Gross', 
    data=plot_data, 
    alpha=0.6,          # Transparency for better visualization
    s=80,               # Size of the points
    edgecolor="black",
    hue='IMDB_Rating',  # Color based on rating
    palette="viridis",
    legend=False
)

plt.title("IMDB Rating vs Gross Revenue", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("IMDB Rating", fontsize=12, labelpad=10)
plt.ylabel("Gross Revenue (in millions)", fontsize=12, labelpad=10)

plt.tight_layout()
plt.show()
```

### **Question 24**
(Visualization) Genres ka count dekhne ke liye top 5 genres ka ek clean Pie Chart banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


all_genres = df['Genre'].dropna().str.split(',').explode().str.strip()
top_5_genres = all_genres.value_counts().head(5)

plt.figure(figsize=(8, 8))
colors = sns.color_palette("pastel")[:5]  # Soft pastel colors for a clean look

plt.pie(
    top_5_genres.values, 
    labels=top_5_genres.index, 
    autopct='%1.1f%%',          # Har slice par percentage dikhane ke liye
    startangle=140,             # Chart ko rotate karne ke liye taake clean dikhe
    colors=colors, 
    textprops={'fontsize': 12, 'fontweight': 'bold'}, # Labels ka style
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}  # Slices ke darmiyan white line
)

centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Top 5 Movie Genres Share", fontsize=16, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

### **Question 25**
(Visualization) Meta_score ka distribution aur outliers check karne ke liye ek Box Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Safe side ke liye Meta_score ko numeric format mein convert karein
df['Meta_score'] = pd.to_numeric(df['Meta_score'], errors='coerce')

# 2. Plot ka size aur clear style set karein
plt.figure(figsize=(10, 5))
sns.set_theme(style="whitegrid")

# 3. Horizontal Box Plot banayein
sns.boxplot(
    x=df['Meta_score'], 
    color="mediumpurple",   # Elegant box color
    flierprops={"marker": "o", "markerfacecolor": "crimson", "markersize": 8}, # Outliers design
    linewidth=2
)

# 4. Strip Plot overlay karein (Optional: real data points ko dots ki shakal mein dekhne ke liye)
sns.stripplot(x=df['Meta_score'], color="black", size=3, alpha=0.3, jitter=0.2)

# 5. Title aur Labels ko design karein
plt.title("Distribution and Outliers of Meta_score", fontsize=16, fontweight='bold', pad=15)
plt.xlabel("Meta score", fontsize=12, labelpad=10)

# 6. Plot display karein
plt.tight_layout()
plt.show()
```

### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Clean the 'Gross' column and convert to numeric
if 'Gross' in df.columns:
    df['Gross'] = df['Gross'].astype(str).str.replace('$', '').str.replace(',', '')
    df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')

# Drop rows where Certificate or Gross is missing for an accurate plot
plot_data = df.dropna(subset=['Certificate', 'Gross'])

# 2. Sort the certificates by their average gross revenue in descending order
# This makes the final bar chart much easier to read and scan
order = plot_data.groupby('Certificate')['Gross'].mean().sort_values(ascending=False).index

# 3. Set the plot size and style
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# 4. Create the Bar Plot
# Seaborn automatically calculates the average (mean) when you pass raw data to barplot
ax = sns.barplot(
    data=plot_data, 
    x='Certificate', 
    y='Gross', 
    order=order, 
    palette="magma",       # Beautiful dark-to-light gradient color scheme
    edgecolor="black",
    hue='Certificate',     # To avoid future pandas/seaborn warnings
    legend=False
)

# 5. Format and design the Titles and Labels
plt.title("Average Gross Revenue by Movie Certificate Category", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Certificate Type", fontsize=12, labelpad=10)
plt.ylabel("Average Gross Earnings (Absolute Value)", fontsize=12, labelpad=10)

# Rotate the X-axis labels slightly if they overlap
plt.xticks(rotation=45, ha='right')

# 6. Display the plot cleanly
plt.tight_layout()
plt.show()
```

### **Question 27**
(Visualization) Har saal (Released_Year) kitni movies release hueen, iska ek Line Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. 'Released_Year' ko numeric format mein convert karein aur missing values hatayein
df['Released_Year'] = pd.to_numeric(df['Released_Year'], errors='coerce')
clean_df = df.dropna(subset=['Released_Year'])

# 2. Har saal ka total movie count nikal kar use saal ke mutabiq sort karein
yearly_counts = clean_df['Released_Year'].value_counts().sort_index()

# 3. Plot ka size aur theme set karein
plt.figure(figsize=(14, 6))
sns.set_theme(style="whitegrid")

# 4. Line Plot draw karein
# marker='o' lagane se har saal ke exact data point par ek chota dot ban jayega
sns.lineplot(
    x=yearly_counts.index, 
    y=yearly_counts.values, 
    color="crimson", 
    linewidth=2.5, 
    marker="o", 
    markersize=5,
    markerfacecolor="black"
)

# 5. Title aur Labels ko design karein
plt.title("Number of Movies Released Per Year", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Release Year", fontsize=12, labelpad=10)
plt.ylabel("Total Movies Count", fontsize=12, labelpad=10)

# X-axis par saal ka gap set karein taaki text aapas mein takraye nahi
plt.xticks(range(int(yearly_counts.index.min()), int(yearly_counts.index.max()) + 1, 5), rotation=45)

# 6. Plot display karein
plt.tight_layout()
plt.show()
```

### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
# 1. Ensure 'No_of_Votes' is numeric (handles any accidental text or commas)
df['No_of_Votes'] = pd.to_numeric(df['No_of_Votes'], errors='coerce')

# 2. Get the top 5 highest voted movies
# ('Series_Title' can be replaced with your actual movie title column name)
top_voted_movies = df.sort_values(by='No_of_Votes', ascending=False).head(5)

# 3. Set the plot size and style
plt.figure(figsize=(12, 5))
sns.set_theme(style="whitegrid")

# 4. Create the Horizontal Bar Chart
ax = sns.barplot(
    data=top_voted_movies,
    x='No_of_Votes',
    y='Series_Title',
    palette="rocket",      # Premium high-contrast color gradient
    edgecolor="black",
    hue='Series_Title',    # To prevent future deprecation warnings
    legend=False
)

# 5. Add exact vote counts as labels at the end of each bar
for container in ax.containers:
    # Formats large numbers with commas (e.g., 2,345,678)
    ax.bar_label(container, fmt='%d', padding=8, fontsize=11, fontweight='bold')

# 6. Format Titles and Labels
plt.title("Top 5 Highest Voted Movies of All Time", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Total Number of Votes", fontsize=12, labelpad=10)
plt.ylabel("Movie Title", fontsize=12, labelpad=10)

# Expand X-axis limit slightly so the text labels don't get clipped
plt.xlim(0, top_voted_movies['No_of_Votes'].max() * 1.15)

# 7. Display the plot cleanly
plt.tight_layout()
plt.show()
```
